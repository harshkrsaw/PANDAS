## Cleaning or Editing Strings

```python
messy_names = pd.Series(['harsh   ,','ayush,  ','rimj:him','  vani  '])
```

```python
messy_names
```

```python
messy_names.str.replace(',','').str.replace(':','').str.strip().str.capitalize()
```

## Alternative with Custom apply() call

```python
def clean_up(name):
    name = name.replace(',','')
    name = name.replace(':','')
    name = name.strip()
    name = name.capitalize()
    return name
```

```python
messy_names.apply(clean_up)
```
