## .merge() method on columns

```python
registrations = pd.DataFrame({'reg_id':[1,2,3,4],'name':['Andrew','Bobo','Claire','David']})
logins = pd.DataFrame({'log_id':[1,2,3,4],'name':['Xavier','Andrew','Yolanda','Bobo']})
```

```python
registrations
```

```python
logins
```

### Inner Join

```python
# Performing inner merge on names means the merge will occur and only the intersection of both the dataframes will show up(intersection of names).

registrations.merge(logins,how = 'inner',on = 'name')
#or (both works same)
pd.merge(registrations,logins,how = 'inner',on = 'name')
```

### Left Join

```python
# Performing a Left Merge on names meanse the merge will occur and all the items from left dataframe will show up and else will NaN

pd.merge(left = registrations,right = logins,how ='left',on = 'name')
```

```python
registrations #Left
```

### Right Join

```python
# Performing a right merge on names means the merge will occur and all the items from right dataframe will occur and else will be NaN

pd.merge(registrations,logins,how ='right',on ='name')
```

```python
logins #right
```

### Outer Join

```python
# Performing a outer merge in names means merge will occur and all the items from both dataframes will show up.

pd.merge(registrations,logins,how='outer',on ='name')
```

## .merge() method on index or rows

```python
registrations = registrations.set_index('name')
```

```python
registrations
```

```python
logins
```

```python
# Now if we need to merge these two where we dont have a common column and want to use the index of one of the columns

pd.merge(registrations,logins,how = 'inner',left_index = True,right_on = 'name')
```

```python
# Changing the sequence of writing.

pd.merge(logins,registrations,how='inner',right_index=True,left_on = 'name')
```

### Dealing with differing key column names in joined tables

```python
registrations = registrations.reset_index()
```

```python
registrations
```

```python
registrations.columns=['reg_name','reg_id']
```

```python
registrations
```

```python
logins
```

```python
# For different column names

pd.merge(registrations,logins,how = 'inner',left_on = 'reg_name',right_on='name')
```

```python
# If want to drop one of the columns 

df = pd.merge(registrations,logins,how = 'inner',left_on = 'reg_name',right_on='name')
df = df.drop('reg_name',axis = 1)
```

```python
df
```

### |Pandas automatically tags duplicate columns

```python
registrations.columns = ['name','id']
```

```python
logins.columns = ['id','name']
```

```python
registrations
```

```python
logins
```

```python
# If the column names are same it automatically gives suffix.

pd.merge(registrations,logins,on='name')
```

```python
# We can keep the suffix by our choice as well

pd.merge(registrations,logins,on='name',suffixes=('_reg','_log'))
```
