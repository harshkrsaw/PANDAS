### .apply() method for multiple columns

```python
#If want to make changes or work with multiple columns with .apply method
```

```python
def quality(total_bill,tip):
    if tip/total_bill > 0.25:
        return 'Good tip'
    else:
        return 'Generous'
```

```python
df[['total_bill','tip']].apply((lambda df:quality(df['total_bill'],df['tip'])),axis = 1)
```

```python
df['tip_quality']=np.vectorize(quality)(df['total_bill'],df['tip'])
```

```python
#Dealing with multilple columns

def temp(bill,tip):
    if bill/4 < tip:
        return 'Generous'
    else: 
        return 'Other'
```

```python
temp(16.99,1.01)
```

```python
#first method

df['Quality_of_tip'] = df[['total_bill','tip']].apply(lambda df:temp(df['total_bill'],df['tip']),axis = 1)
```

```python
#Second method (Faster)

df['Quality_of_tip'] = np.vectorize(temp)(df['total_bill'],df['tip'])
```
