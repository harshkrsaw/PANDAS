### HTML Files

```python
# We can bring any url from internet and read the html tables here and work on.... Only if they havent encripted their website.

url = 'https://worldpopulationreview.com/countries/india'
```

```python
# We can grab all the tables from the link

tables = pd.read_html(url)
```

```python
len(tables)
```

```python
table = tables[1]
```

```python
table
```

```python
table.columns
```

```python
table = table.set_index(table['Rank'])
```

```python
table = table.drop('Rank',axis = 1)
```

```python
table
```

```python
table.columns = ['Rank','Sheher','2026 pop.','share']
```

```python
table.reset_index()
```

```python
# And we can save it from here in any format or name.

table.to_html('sample_india_pop.html')
```

### SQL connection

```python
from sqlalchemy import create_engine
```

```python
temp_db = create_engine('sqlite:///:memory:')
```

```python
df = pd.DataFrame(np.random.randint(0,100,(4,4)),columns =['a','b','c','d'])
```

```python
df
```

```python
df.to_sql('new_table',con =temp_db)
```
