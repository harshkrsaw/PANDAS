```python
import pandas as pd
import numpy as np
```

## pd.pivot() and pd.pivot_table()

```python
help(pd.pivot)
```

```python
df = pd.read_csv('Sales_Funnel_CRM.csv')
```

```python
df
```

```python
pd.pivot(data =df, index = 'Company',columns = 'Product',values= ['Licenses','Sale Price'])
```

```python
# We use it for making these type of pivot tables.

df.pivot(index = ['Company'],columns = ['Product'],values = ['Licenses'])
```

```python
df.groupby('Contact').mean(numeric_only=True)['Sale Price']
```

```python
# If we want to use some of aggregation function we use this.

pd.pivot_table(df,index = ['Company'],aggfunc='sum',values =['Sale Price','Licenses'] )
```

```python
# We can do the same thing by .groupby() as well

df.groupby(['Company']).sum(numeric_only = True)[['Licenses','Sale Price']]
```

```python
df.head()
```

```python
# Multilevel index

pd.pivot_table(df,index = ['Account Manager','Contact'],values = ['Sale Price','Licenses'],aggfunc='sum')
```

```python
# Same doing by .groupby()

df.groupby(['Account Manager','Contact']).sum(numeric_only = True)[['Licenses','Sale Price']]
```

```python
# Multilevel index with multilevel column

pd.pivot_table(df,index = ['Account Manager','Contact'],columns = ['Product'],values = ['Sale Price'],aggfunc='sum')
```

```python
# We can fill the Nan by adding fill_value parameter

pd.pivot_table(df,index = ['Account Manager','Contact'],columns = ['Product'],values = ['Sale Price'],aggfunc='sum',fill_value=0)
```

```python
# We can add multipe aggfunc as well 

pd.pivot_table(df,index = ['Account Manager','Contact'],columns = ['Product'],values = ['Sale Price'],aggfunc=['sum','mean'],fill_value=0)
```

```python
# Now the table look too much messy and its really not suggest to make it this much messy.. you can try do things one at time.

pd.pivot_table(df,index = ['Account Manager','Contact','Product'],values = ['Sale Price'],aggfunc=['sum'],fill_value=0)
```

```python
# Margine parameter(it just sum up all the values at the end)

pd.pivot_table(df,index = ['Account Manager','Contact','Product'],values = ['Sale Price'],aggfunc=['sum'],fill_value=0,margins = True)
```
